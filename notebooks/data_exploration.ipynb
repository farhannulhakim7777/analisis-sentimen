```python
# Cell 1: Impor, Magic Command, dan Latih Model
# Magic Command ini WAJIB agar plot tampil di Notebook
%matplotlib inline 

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np # <--- PINDAHKAN INI DI SINI
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

# Impor fungsi dari main.py (pastikan main.py sudah disimpan!)
from main import create_mock_data, train_sentiment_model 

# Dapatkan data dan latih model
df_data = create_mock_data()
model, vectorizer, X_test, y_test = train_sentiment_model(df_data)

print("--- STATUS ---")
print("Data uji X_test dan y_test sudah siap untuk dievaluasi.")
print("Model Naive Bayes telah dilatih.")
```

```python
# Cell 2: Prediksi dan Evaluasi

# Prediksi menggunakan data uji (X_test)
X_test_vectorized = vectorizer.transform(X_test)
y_pred = model.predict(X_test_vectorized)

# Cetak Laporan Klasifikasi
print("--- Laporan Klasifikasi ---")
print(classification_report(y_test, y_pred))

# Cetak Akurasi
print(f"Akurasi Model: {accuracy_score(y_test, y_pred):.2f}")
```

```python
# Cell 3: Visualisasi Confusion Matrix
from sklearn.metrics import confusion_matrix # Import ulang untuk memastikan definisi cm

# Hitung Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

# Siapkan plot
plt.figure(figsize=(5, 5))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix Sentimen')
plt.colorbar()

# Atur label dan tambahkan angka di dalam kotak (Menggunakan np yang sudah diimpor)
tick_marks = [0, 1]
plt.xticks(tick_marks, ['Negatif (0)', 'Positif (1)'])
plt.yticks(tick_marks, ['Negatif (0)', 'Positif (1)'])
plt.ylabel('Label Sebenarnya (True Label)')
plt.xlabel('Label Prediksi (Predicted Label)')

# Tambahkan angka di tengah kotak
thresh = cm.max() / 2.
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, format(cm[i, j], 'd'),
                 ha="center", va="center",
                 color="white" if cm[i, j] > thresh else "black")

plt.show() # Perintah ini menampilkan plot
```
